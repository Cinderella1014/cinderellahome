```python
import os
import pandas as pd

# 设置输入文件夹路径
input_folder = "C:/Users/86183/BS/BS/stonedata/combined.csv" # 存放txt文件的文件夹
output_file = "last_loss_data.csv"  # 合并后的csv文件路径

# 创建一个空的DataFrame来存储最后一列的数据
last_column_data = pd.DataFrame()

# 遍历输入文件夹中的所有文件
for filename in os.listdir(input_folder):
    if filename.endswith(".txt"):  # 确保只处理txt文件
        # 构建完整的文件路径
        txt_path = os.path.join(input_folder, filename)
        
        try:
            # 读取txt文件
            data = pd.read_csv(txt_path, sep='\t', header=None)  # 假设txt文件没有表头，根据实际分隔符调整sep
            
            # 提取最后一列的数据
            last_column = data.iloc[:, -1]
            
            # 将最后一列数据添加到DataFrame中
            last_column_data[filename] = last_column
            print(f"成功提取最后一列数据: {filename}")
        except Exception as e:
            print(f"提取失败: {filename}, 错误: {e}")

# 将提取的数据保存为csv文件
last_column_data.to_csv(output_file, index=False)
print(f"所有文件的最后一列数据已合并到: {output_file}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter

# 1. 模拟生成拉曼光谱数据（示例：含噪声的尿路结石光谱）
x = np.linspace(500, 2000, 1000)  # 拉曼位移（cm⁻¹）
y_true = 10 * np.exp(-(x - 1200)**2 / 1000)  # 真实信号（假设为高斯峰）
noise = np.random.normal(0, 0.5, len(x))      # 高斯噪声
y_noisy = y_true + noise                       # 含噪声信号

# 2. 应用SG滤波去噪
window_length = 21    # 滑动窗口大小（必须为奇数）
polyorder = 3         # 多项式阶数
y_smoothed = savgol_filter(y_noisy, window_length, polyorder)

# 3. 绘制结果对比
plt.figure(figsize=(10, 6))
plt.plot(x, y_noisy, 'b-', alpha=0.5, label='原始含噪信号')
plt.plot(x, y_smoothed, 'r-', linewidth=2, label='SG滤波后')
plt.plot(x, y_true, 'k--', alpha=0.7, label='真实信号（参考）')
plt.xlabel('拉曼位移 (cm⁻¹)')
plt.ylabel('强度 (a.u.)')
plt.title('尿路结石拉曼光谱的SG去噪效果')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
def sg_denoise(spectra, window_length=21, polyorder=3):
    """
    参数说明：
    - window_length: 滑动窗口大小（需为奇数，建议15-25）‌:ml-citation{ref="3" data="citationList"}
    - polyorder: 多项式阶数（尿路结石推荐3阶）‌:ml-citation{ref="3" data="citationList"}
    """
    assert window_length % 2 == 1, "窗口长度必须是奇数"
    return np.array([savgol_filter(s, window_length, polyorder) for s in spectra])

# 执行批量去噪
denoised_spectra = sg_denoise(spectra)  # 默认参数‌:ml-citation{ref="3,7" data="citationList"}
import pandas as pd
#pd.DataFrame(denoised_spectra).to_csv('denoised_spectra.csv', index=False, header=False)  # ‌:ml-citation{ref="2,3" data="citationList"}
```
